# 从排名到贡献与成本

固定同一资本和权重顺序，逐资产计算调仓。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

按排名构建每侧两只的等权组合。

```python
names=np.array(list('ABCDE'));old_rank=np.array([0,1,2,3,4]);new_rank=np.array([4,0,2,1,3])
old=np.zeros(5);new=np.zeros(5)
old[old_rank[:2]]=.5;old[old_rank[-2:]]=-.5
new[new_rank[:2]]=.5;new[new_rank[-2:]]=-.5
orders=new-old;print('asset, old, new, order:')
for i in range(5):print(names[i],old[i],new[i],orders[i])
```

以新持仓计算下一期贡献，忽略调仓前权重漂移。

```python
returns=np.array([.03,.01,0,-.01,-.02]);turnover=abs(orders).sum();cost_rate=.001
contributions=new*returns;gross=contributions.sum();cost=cost_rate*turnover
print('contributions:',contributions,'gross, turnover, cost, net:',gross,turnover,cost,gross-cost)
rates=np.linspace(0,.004,100);plt.plot(rates*10000,gross-rates*turnover);plt.axhline(0,color='gray');plt.xlabel('cost (bp)');plt.show()
```
