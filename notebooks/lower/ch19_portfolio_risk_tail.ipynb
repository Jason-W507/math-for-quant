# 离散 ES 与反向压力

尾部质量可能只取边界观测的一部分。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

从最坏损失向下分配尾部质量，显式保留每个观测的份额。

```python
losses=np.array([0.,1.,2.,10.]);alpha=.6;ordered=np.sort(losses)[::-1];remaining=1-alpha;total=0.
for value in ordered:
    mass=min(1/len(losses),remaining);total+=mass*value;remaining-=mass
    print('loss, tail mass:',value,mass)
print('ES:',total/(1-alpha),'VaR:',np.sort(losses)[int(np.ceil(alpha*len(losses)))-1])
```

沿同一压力方向求全部根，再选第一次穿越。

```python
threshold=10.;roots=np.roots([-.384,4.4,-threshold]);valid=sorted(r.real for r in roots if abs(r.imag)<1e-10 and r.real>=0)
s=np.linspace(0,12,200);loss=4.4*s-.384*s*s
plt.plot(s,loss);plt.axhline(threshold,color='red');plt.xlabel('stress scale');plt.ylabel('loss');plt.show();print('roots:',valid,'first:',valid[0])
```
