# Ridge、lasso 与共线性

惩罚目标写清以后，每个系数都能逐步计算。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

一列设计同时比较三种解。

```python
x=np.array([1.,2.]);y=np.array([1.,1.]);lam=1.;xx=x@x;xy=x@y
ols=xy/xx;ridge=xy/(xx+lam);lasso=np.sign(xy)*max(abs(xy)-lam,0)/xx
print('OLS, ridge, lasso:',ols,ridge,lasso)
```

共线特征使小奇异方向敏感；画出惩罚路径。

```python
x=rng.normal(size=100);X=np.column_stack([x,x+.01*rng.normal(size=100)]);y=x+.2*rng.normal(size=100)
lambdas=np.logspace(-5,2,100);coef=np.array([np.linalg.solve(X.T@X+l*np.eye(2),X.T@y) for l in lambdas])
plt.semilogx(lambdas,coef);plt.xlabel('lambda');plt.ylabel('coefficient');plt.show()
print('singular values:',np.linalg.svd(X,compute_uv=False))
```
