# 递推记忆与因果注意力

观察不同位置贡献，再遮蔽未来位置。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

线性递推展示记忆衰减与梯度的同一几何权重。

```python
inputs=np.array([1.,0.,0.,0.,0.]);a=.5;state=0.;states=[]
for value in inputs:state=a*state+value;states.append(state)
print('states:',states,'influence of first input:',a**np.arange(len(inputs)))
```

注意力先生成得分，再应用掩码、稳定 softmax 和加权平均。

```python
Q=np.ones((3,1));K=np.array([[0.],[1.],[2.]]);V=np.array([[10.],[20.],[1000000.]])
scores=Q@K.T/np.sqrt(K.shape[1]);mask=np.triu(np.ones((3,3),dtype=bool),1)
masked=np.where(mask,-np.inf,scores);exponent=np.exp(masked-np.max(masked,axis=1,keepdims=True));weights=exponent/exponent.sum(axis=1,keepdims=True);output=weights@V
print('scores:\n',scores,'mask:\n',mask,'weights:\n',weights,'output:\n',output)
plt.imshow(weights,vmin=0,vmax=1);plt.xlabel('key position');plt.ylabel('query position');plt.colorbar();plt.show()
```
