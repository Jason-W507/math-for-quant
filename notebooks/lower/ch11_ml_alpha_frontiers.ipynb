```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

均值邻域传播相当于一个行随机矩阵。

```python
A=np.array([[.5,.5,0],[1/3,1/3,1/3],[0,.5,.5]]);h=np.array([1.,2.,4.]);history=[h.copy()]
for layer in range(30):h=A@h;history.append(h.copy())
print('first layer:',history[1],'limit:',h,'analytic:',16/7)
plt.plot(history);plt.xlabel('layer');plt.ylabel('node value');plt.show()
```

动作集合决定最优：第一期最多买 1，第二期最多买 2。

```python
quantity=2;paths=[]
for first in [0,1]:
    second=quantity-first;cost=first+3*second;paths.append((cost,first,second))
print('all paths:',paths,'best:',min(paths))
```
