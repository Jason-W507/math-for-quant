# 状态、不变量与构造

先枚举小问题，再将规律写成覆盖所有规模的证明。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

从剩余硬币数建立胜负递推，允许取 1 或 2。

```python
winning=[False]*21;move=[None]*21
for n in range(1,21):
    for a in [1,2]:
        if a<=n and not winning[n-a]:winning[n]=True;move[n]=a;break
print('losing states:',[n for n in range(21) if not winning[n]])
```

逐位匹配开关，后续操作不再影响已匹配的前缀。

```python
target=np.array([1,0,1,1,0,1]);state=np.zeros(len(target),dtype=int);actions=[]
for i in range(len(target)-1):
    if state[i]!=target[i]:state[i:i+2]^=1;actions.append(i)
print('actions:',actions,'result:',state,'reachable:',np.array_equal(state,target))
```

生日配对期望与至少一对概率不同。

```python
for n in [10,23,50]:
    expected=n*(n-1)/(2*365);probability=1-np.prod(1-np.arange(n)/365)
    print(n,expected,probability)
```
