# 概率阈值、仓位与成交

把预测含义转成条件期望收益，再施加成本与风险。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

不对称盈亏下的概率阈值来自条件期望。

```python
gain=.02;loss=.01;cost=.002;p=np.linspace(0,1,101)
expected=p*gain-(1-p)*loss-cost;threshold=(loss+cost)/(gain+loss)
plt.plot(p,expected);plt.axhline(0,color='gray');plt.axvline(threshold);plt.xlabel('probability');plt.ylabel('expected net return');plt.show();print('threshold:',threshold)
```

两期目标变化与部分成交逐资产计算。

```python
targets=np.array([[1.,0,-1],[-1,1,0]]);fills=np.array([[1.,1,1],[.5,1,0]]);returns=np.array([[.02,0,-.01],[0,.01,-.01]])
position=np.zeros(3);gross=0.;turnover=0.
for target,fill,ret in zip(targets,fills,returns):
    order=target-position;trade=fill*order;position+=trade;turnover+=abs(trade).sum();gross+=position@ret
    print('orders:',order,'trades:',trade,'positions:',position,'contribution:',position@ret)
print('gross, turnover, net:',gross,turnover,gross-.001*turnover)
```
