# 执行日程与库存反馈

从同一个成本函数出发比较连续解与整数动作。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

枚举整数两期日程，显示冲击与风险的权衡。

```python
Q=6;eta=1.;phi=2.;first=np.arange(Q+1);costs=eta*(first**2+(Q-first)**2)+phi*(Q-first)**2
print('first slice, objective:',np.column_stack([first,costs]));print('continuous optimum:',(eta+phi)*Q/(2*eta+phi))
plt.plot(first,costs,'o-');plt.xlabel('first slice');plt.ylabel('objective');plt.show()
```

有限状态动态规划从终端向前算，不隐藏状态转移。

```python
periods=3;quantity=6;cap=3;eta=1.;risk=.5
V=np.full((periods+1,quantity+1),np.inf);V[-1,0]=0;policy=np.zeros((periods,quantity+1),dtype=int)
for t in range(periods-1,-1,-1):
    for remaining in range(quantity+1):
        candidates=[(eta*a*a+risk*(remaining-a)**2+V[t+1,remaining-a],a) for a in range(min(cap,remaining)+1)]
        V[t,remaining],policy[t,remaining]=min(candidates)
remaining=quantity;schedule=[]
for t in range(periods):a=policy[t,remaining];schedule.append(a);remaining-=a
print('schedule, objective, remainder:',schedule,V[0,quantity],remaining)
```

每次成交后现金和库存共同决定财富。

```python
mid=100.;half=.5;kappa=.1;inventory=0;cash=0.
for side in ['bid','bid','ask']:
    center=mid-kappa*inventory;bid=center-half;ask=center+half
    if side=='bid':cash-=bid;inventory+=1
    else:cash+=ask;inventory-=1
    print(side,'quotes:',bid,ask,'inventory:',inventory,'wealth:',cash+inventory*mid)
```
