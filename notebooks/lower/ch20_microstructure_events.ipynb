# 事件率、Hawkes 与 FIFO

事件次数与耗尽数量只有在一笔一手的假设下相同。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

Poisson 队列概率来自等待耗尽的手数。

```python
mean=3.;ahead=np.arange(8);quantity=1
probability=stats.poisson.sf(ahead+quantity-1,mean)
print('ahead, full-fill probability:',np.column_stack([ahead,probability]))
plt.plot(ahead,probability,'o-');plt.xlabel('quantity ahead');plt.ylabel('fill probability');plt.show()
```

有限历史 Hawkes 在事件前计算强度，再把新事件加入历史。

```python
events=np.array([.2,.7,1.4,2.]);T=2.5;mu=.8;alpha=.3;beta=1.2;initial=.1;intensities=[]
for i,t in enumerate(events):
    intensity=mu+initial*np.exp(-beta*t)+alpha*np.exp(-beta*(t-events[:i])).sum();intensities.append(intensity)
compensator=mu*T+initial*(1-np.exp(-beta*T))/beta+alpha/beta*np.sum(1-np.exp(-beta*(T-events)))
print('pre-event intensities:',intensities,'log likelihood:',np.log(intensities).sum()-compensator,'stationary mean:',mu/(1-alpha/beta))
```

同价两张订单按到达次序消耗，未成交量留在订单中。

```python
queue=[3,2];market=4;fills=[]
for i in range(len(queue)):
    filled=min(market,queue[i]);queue[i]-=filled;market-=filled;fills.append(filled)
print('fills:',fills,'remaining queue:',queue,'unfilled market order:',market)
```
