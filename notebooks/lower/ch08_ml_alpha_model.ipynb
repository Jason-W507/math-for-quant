# 树桩与逐轮提升

枚举每个训练划分，显示叶均值和误差，再拟合残差。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

排序后相邻值之间的任意阈值给相同训练划分。

```python
x=np.array([-2.,-1.,1.,3.]);y=np.array([-4.,-2.,1.,5.]);thresholds=(x[:-1]+x[1:])/2
for s in thresholds:
    left=x<=s;prediction=np.where(left,y[left].mean(),y[~left].mean())
    print('threshold, leaves, SSE:',s,y[left].mean(),y[~left].mean(),np.sum((y-prediction)**2))
```

每轮显式计算负梯度、搜索树桩并更新。

```python
prediction=np.full(len(y),y.mean());rate=.2;losses=[]
for step in range(20):
    residual=y-prediction;candidates=[]
    for s in thresholds:
        left=x<=s;tree=np.where(left,residual[left].mean(),residual[~left].mean())
        candidates.append((np.sum((residual-tree)**2),tree))
    _,tree=min(candidates,key=lambda pair:pair[0]);prediction+=rate*tree;losses.append(np.mean((y-prediction)**2))
print('fitted values:',prediction);plt.plot(losses);plt.xlabel('boosting round');plt.ylabel('training MSE');plt.show()
```
