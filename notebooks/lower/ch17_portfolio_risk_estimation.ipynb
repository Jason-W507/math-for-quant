# 协方差谱、收缩与贡献

半正定来自平方和结构，正定收缩来自额外结构假设。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

中心化后直接构造样本协方差，观察高维秩。

```python
returns=rng.normal(size=(5,8));centered=returns-returns.mean(axis=0);cov=centered.T@centered/4
print('eigenvalues:',np.linalg.eigvalsh(cov),'rank:',np.linalg.matrix_rank(cov))
```

一个精确奇异矩阵在收缩后抬高零特征值。

```python
sample=np.ones((2,2));lambdas=np.linspace(.01,1,100)
condition=[]
for lam in lambdas:
    estimate=(1-lam)*sample+lam*np.eye(2);condition.append(np.linalg.cond(estimate))
plt.plot(lambdas,condition);plt.xlabel('shrinkage');plt.ylabel('condition number');plt.show()
```

先计算方差贡献，再除以总波动得到波动贡献。

```python
vol=np.array([.2,.4]);weights=(1/vol)/np.sum(1/vol);cov=np.diag(vol**2)
contribution=weights*(cov@weights);total=weights@cov@weights
print('weights, variance contributions, total:',weights,contribution,total)
print('volatility contributions:',contribution/np.sqrt(total))
```
