# 观点后验与情景尾部优化

先计算观点的精度加权，再枚举小组合。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

标量后验将观点置信度直接写成权重。

```python
prior=.04;view=.08;v=.0025
for omega in [v*10,v,v/3]:
    mean=(omega*prior+v*view)/(v+omega);variance=v*omega/(v+omega)
    print('view variance, posterior mean, variance:',omega,mean,variance)
```

八个情景在 75% 水平取最坏两个损失；权重限制直接进入枚举。

```python
scenarios=np.array([[-.06,-.01],[-.03,-.02],[.01,-.005],[.02,.01],[.04,.02],[-.01,.005],[.03,.015],[0,0]])
weights=np.linspace(.2,.8,121);objectives=[]
for first in weights:
    w=np.array([first,1-first]);loss=-scenarios@w;objectives.append(np.sort(loss)[-2:].mean())
index=np.argmin(objectives);print('best weight:',weights[index],1-weights[index],'CVaR:',objectives[index])
plt.plot(weights,objectives);plt.xlabel('first asset weight');plt.ylabel('CVaR');plt.show()
```

为何把权重区间扩成 [0,1] 会改变结果？

## 反馈

这是改变可行域，不能把前后结果差异当作求解器误差；在本组情景下会更集中于第二资产。

参数改变后应重新解释结果，不要求复现某次随机实验的小数。
