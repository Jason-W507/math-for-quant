# 从平均数到加权平均

三期收益分别为 1%、−2%、3%。怎样用一个数概括它们？先求算术平均，再改变每期的权重，观察平均值的位置。本例计算的是各期收益的加权摘要，不是多期复利收益。

```python
import numpy as np

returns = np.array([0.01, -0.02, 0.03])
arithmetic_mean = returns.mean()
print("算术平均：", arithmetic_mean)
print("三个贡献：", returns / len(returns))
print("与解析值 1/150 的差：", arithmetic_mean - 1/150)
```

三个收益的和为 0.02，除以 3 得到 $1/150$，约为 0.667%。算术平均等价于使用相同权重。下面提高第三期的权重：在运行前，先判断平均值会增大还是减小。

```python
weights = np.array([0.25, 0.25, 0.50])
contributions = weights * returns
weighted_mean = contributions.sum()
print("权重和：", weights.sum())
print("各期贡献：", contributions)
print("加权平均：", weighted_mean)
print("最小与最大收益：", returns.min(), returns.max())
```

加权平均为 0.0125，位于最小收益与最大收益之间。若权重非负且和为 1，每项都在区间内，这种加权平均也在区间内；第一章给出一般证明。允许负权重后，这个结论还成立吗？

```python
signed_weights = np.array([0., -1., 2.])
print("新权重和：", signed_weights.sum())
print("新加权值：", signed_weights @ returns)
```

新权重仍然和为 1，加权值却是 0.08，超过了最大收益 0.03。这个反例说明，权重归一化本身不足以保证上下界。
