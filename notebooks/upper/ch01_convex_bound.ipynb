# 凸组合与量词顺序

先计算三个资产的贡献，再观察预先固定权重与事后选择权重的区别。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

权重和为 1 且非负时，收益位于各分量的最小值与最大值之间。修改下面两个向量。

```python
returns=np.array([.02,-.01,.03]); weights=np.array([.5,.3,.2])
contributions=weights*returns
print('contributions, sum, bounds:', contributions, contributions.sum(), returns.min(), returns.max())
short_weights=np.array([1.5,-.5,0]); print('short portfolio:', short_weights@returns)
```

两种互相相反的情形：每种情形都能事后获利，但预先选同一个权重能否保证正收益？

```python
w=np.linspace(0,1,201)
r1=2*w-1; r2=1-2*w
plt.plot(w,r1,label='scenario 1');plt.plot(w,r2,label='scenario 2');plt.plot(w,np.minimum(r1,r2),label='worst case')
plt.xlabel('weight');plt.ylabel('return');plt.legend();plt.show()
```

两期收益的方差包含两份协方差。

```python
variance=4.; covariance=1.
print('sum variance:',2*variance+2*covariance,'independent assumption:',2*variance)
```
