# 误差范数与求和次序

同一尖峰可以在一种范数下消失，在另一种范数下保留质量。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

解析范数来自高度的 p 次方乘宽度。改变高度指数 a。

```python
n=np.geomspace(1,10000,100);a=1/3
for p in [1,2,3,4]:plt.loglog(n,n**(a-1/p),label=f'p={p}')
plt.xlabel('n');plt.ylabel('Lp norm');plt.legend();plt.show()
```

逐项构造含正负项的矩阵，比较不同扩张边界。

```python
for n in [3,10,100]:
    row=np.arange(1,n+2)[:,None];col=np.arange(1,n+1)[None,:]
    matrix=(row==col).astype(int)-(row==col+1).astype(int)
    print(n,'square:',matrix[:n].sum(),'rectangle:',matrix.sum(),'absolute square:',abs(matrix[:n]).sum())
```
