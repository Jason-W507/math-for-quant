# Markov 递推与命中问题

既计算分布传播，也计算退出概率和时长。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

矩阵乘法与标量递推回答同一个两状态问题。

```python
P=np.array([[.8,.2],[.3,.7]]);mu=np.array([1.,0.])
for n in range(6):
    print(n,mu,'scalar:',.6+.4*.5**n);mu=mu@P
```

有限状态命中方程是线性方程；时长方程每一步多加 1。

```python
N=5;Q=np.zeros((N-1,N-1))
for i in range(N-1):
    if i>0:Q[i,i-1]=.5
    if i<N-2:Q[i,i+1]=.5
rhs=np.zeros(N-1);rhs[-1]=.5
hit=np.linalg.solve(np.eye(N-1)-Q,rhs);duration=np.linalg.solve(np.eye(N-1)-Q,np.ones(N-1))
print('hit:',hit,'duration:',duration)
print('analytic:',np.arange(1,N)/N,np.arange(1,N)*(N-np.arange(1,N)))
```

有界停止时只累加停止前的公平增量。

```python
positions=np.full(5000,2);times=np.zeros(5000,dtype=int)
for step in range(300):
    active=(positions>0)&(positions<N)
    positions[active]+=rng.choice([-1,1],active.sum());times[active]+=1
print('empirical exit probability, mean duration:',(positions==N).mean(),times.mean())
```
