# 约束最优与软阈值

通过小问题看出可行域和不可微点怎样改变答案。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

约束 x+y≥b 且非负时，b≤0 与 b>0 属于不同活跃集。

```python
b=np.linspace(-1,2,101);x=np.maximum(b,0)/2;primal=x*x;dual=np.maximum(b,0)/2
plt.plot(b,x,label='x=y');plt.plot(b,primal,label='optimal objective');plt.legend();plt.xlabel('b');plt.show()
```

分段计算软阈值，再直接评价目标函数。

```python
a=.8;lam=.3;grid=np.linspace(-1,1.5,501)
answer=np.sign(a)*max(abs(a)-lam,0)
loss=.5*(grid-a)**2+lam*abs(grid)
plt.plot(grid,loss);plt.axvline(answer,color='red');plt.xlabel('x');plt.ylabel('objective');plt.show()
print('soft threshold:',answer,'grid minimum:',grid[np.argmin(loss)])
```
