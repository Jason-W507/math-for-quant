# 简单函数、尖峰与截断期望

用有限层的高度乘质量理解积分，再比较两种不同的尾部。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

直接求每个台阶的高度与宽度乘积，不把网格点平均误认为一般积分定义。

```python
for m in [2,4,8]:
    count=2**m; heights=np.arange(count)/count; width=1/count
    integral=np.sum(heights*width)
    print(m,integral,'analytic:',.5-1/(2*count))
```

尖峰高度 n、宽度 1/n，面积保持 1；固定点最终看不见它。

```python
n=np.array([10,100,1000]);point=.2
print('height, width, area, value at .2:\n',np.column_stack([n,1/n,n*(1/n),n*((point>0)&(point<=1/n))]))
```

对 Pareto 指数 1 和 2，计算截断随机变量 min(X,n) 的期望。

```python
cap=np.geomspace(1,10000,200)
plt.semilogx(cap,1+np.log(cap),label='tail index 1');plt.semilogx(cap,2-1/cap,label='tail index 2')
plt.xlabel('truncation cap');plt.ylabel('expected truncated value');plt.legend();plt.show()
```
