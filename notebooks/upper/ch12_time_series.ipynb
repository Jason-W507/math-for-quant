# AR 预测与 Kalman 更新

先看已知参数下的预测误差，再逐步融合有噪观测。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

每个未来创新都贡献预测误差方差。

```python
phi=.8;variance=1.;current=1.;h=np.arange(1,11)
forecast=phi**h*current;error_var=np.array([variance*sum(phi**(2*j) for j in range(k)) for k in h])
plt.plot(h,forecast);plt.fill_between(h,forecast-1.96*np.sqrt(error_var),forecast+1.96*np.sqrt(error_var),alpha=.2)
plt.xlabel('horizon');plt.ylabel('forecast and model interval');plt.show()
```

局部水平模型逐行显示先验、创新、增益和后验。改变 Q 或 R。

```python
Q=.25;R=1.;m=0.;P=1.
for observation in [1.,-.5,.25]:
    prior_m=m;prior_P=P+Q;innovation=observation-prior_m;K=prior_P/(prior_P+R)
    m=prior_m+K*innovation;P=(1-K)*prior_P
    print('prior, innovation, gain, posterior:',prior_m,innovation,K,m,P)
```
