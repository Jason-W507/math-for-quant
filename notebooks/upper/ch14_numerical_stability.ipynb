分别改变公式和输入，区分算法误差与问题敏感性。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

共轭式保留小差值；高精度给出独立参照。

```python
from decimal import Decimal, localcontext
x=1e16;direct=np.sqrt(x+1)-np.sqrt(x);stable=1/(np.sqrt(x+1)+np.sqrt(x))
with localcontext() as ctx:
    ctx.prec=70;xd=Decimal('1e16');exact=(xd+1).sqrt()-xd.sqrt()
print('direct, stable, reference:',direct,stable,exact)
```

相减两条方程可知精确扰动为 delta/epsilon。

```python
for epsilon in [1e-2,1e-4,1e-8]:
    A=np.array([[1.,1.],[1.,1+epsilon]]);truth=np.ones(2);b=A@truth;changed=b.copy();changed[1]+=1e-10
    estimated=np.linalg.solve(A,changed)
    print(epsilon,'solution:',estimated,'cond:',np.linalg.cond(A),'residual original:',np.linalg.norm(A@estimated-b),'residual changed:',np.linalg.norm(A@estimated-changed))
```
