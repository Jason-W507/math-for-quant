# 压缩迭代与一致收敛

误差、步长与全域上界是三个不同的量。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

先逐步迭代，保留每一步与已知不动点的距离。

```python
q=.8; target=.1; x=0.; rows=[]
for n in range(1,26):
    new=(1-q)*target+q*x
    rows.append((n,new,abs(new-x),abs(new-target),q/(1-q)*abs(new-x)))
    x=new
rows=np.array(rows);print('n, value, step, error, bound:\n',rows[[0,4,20]])
```

改变 q 后重新运行上一格。q 接近 1 时，步长可以远小于剩余误差。

```python
plt.semilogy(rows[:,0],rows[:,2],label='step');plt.semilogy(rows[:,0],rows[:,3],label='error')
plt.xlabel('iteration');plt.legend();plt.show()
needed=int(np.ceil(np.log(.001/target)/np.log(q)));print('iterations needed:',needed)
```

固定点上的 x^n 衰减，不代表整个 [0,1] 上一致收敛。移动点始终保留 1/2 的误差。

```python
n=np.array([2,10,100,1000]); witness=2.**(-1/n)
print('n, fixed value, moving value:\n',np.column_stack([n,.5**n,witness**n]))
```
