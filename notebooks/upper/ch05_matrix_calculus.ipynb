# 梯度、曲率与有限差分

先从微分写出梯度，再比较差分。中心差分返回零不一定意味着可微。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

非对称二次型必须先对称化。

```python
A=np.array([[4.,2.],[0.,3.]]);b=np.array([-1.,2.]);x=np.array([.5,-1.])
def objective(x):return .5*x@A@x+b@x
gradient=(A+A.T)@x/2+b
print('value, gradient, Hessian:',objective(x),gradient,(A+A.T)/2)
for h in [1e-2,1e-5,1e-8,1e-12]:
    finite=np.array([(objective(x+h*e)-objective(x-h*e))/(2*h) for e in np.eye(2)])
    print(h,finite,np.max(abs(finite-gradient)))
```

用指数函数观察截断与舍入的竞争；二次型没有相同的截断项。

```python
h=np.logspace(-1,-16,80);error=abs((np.exp(h)-np.exp(-h))/(2*h)-1)
plt.loglog(h,error);plt.xlabel('step');plt.ylabel('derivative error');plt.show()
h0=.001;print('absolute value: left, right, central:',-abs(-h0)/h0,abs(h0)/h0,(abs(h0)-abs(-h0))/(2*h0))
```
