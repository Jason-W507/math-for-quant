# 时间边界与成本敏感性

用明确时间和固定资本贡献分析研究结论。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

标签采用闭区间 [t,t+4]，与验证标签交集决定删除哪些训练样本。

```python
train=np.arange(1,21);validation=np.arange(21,26);h=5
overlap=np.array([any(max(t,v)<=min(t+h-1,v+h-1) for v in validation) for t in train])
print('removed:',train[overlap],'retained:',train[~overlap])
```

净收益随成本率改变，不等于资金容量曲线。

```python
gross=.003;turnover=2.;other=.0004;c=np.linspace(0,.003,100)
net=gross-turnover*c-other
plt.plot(c*10000,net);plt.axhline(0,color='gray');plt.xlabel('cost per turnover (bp)');plt.ylabel('net contribution');plt.show()
print('break even:',(gross-other)/turnover)
```
