# 方差缩减与 bootstrap

每种抽样方法先写清每条路径对同一目标的贡献。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

普通样本与已知均值为 1/2 的控制变量共享 U。

```python
N=20000;u=rng.random(N);plain=u*u;control=u*u-u+.5
importance=np.sqrt(rng.random(N))/2
for name,contribution in [('plain',plain),('control',control),('importance',importance)]:
    print(name,'mean:',contribution.mean(),'se:',contribution.std(ddof=1)/np.sqrt(N))
print('analytic variances:',4/45,1/180,1/72)
```

穷举所有有序重抽样，可精确计算条件 bootstrap 方差。

```python
from itertools import product
data=[1,2,4];means=np.array([np.mean(row) for row in product(data,repeat=3)])
print('states, mean, variance:',len(means),means.mean(),means.var(),'reference:',14/27)
plt.hist(means,bins=np.linspace(.8,4.2,18));plt.xlabel('bootstrap mean');plt.show()
```
