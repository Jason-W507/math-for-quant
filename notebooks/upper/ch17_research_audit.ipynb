# 三笔交易能支持什么结论

逐笔计算贡献、成本和成交压力，再区分算术与统计证据。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

以下数值都按同一固定资本计量。

```python
gross=np.array([.02,-.01,.03]);cost_types=np.array([.0005,.0007,.0008]);fills=np.ones(3)
cost=np.full(3,cost_types.sum())*fills;net=gross*fills-cost
print('gross, cost, net:',gross*fills,cost,net,'total:',net.sum())
print('if instead sequential returns:',np.prod(1+gross)-1)
```

只改变最后一笔成交，保留线性成本假设。

```python
fills[-1]=.5;net_half=(gross-cost_types.sum())*fills
print('partial-fill contribution:',net_half,'total:',net_half.sum())
pressure=np.linspace(0,8,100);plt.plot(pressure,gross.sum()-3*cost_types.sum()*pressure);plt.axhline(0,color='gray');plt.xlabel('cost multiplier');plt.show()
```

p 值是另外提供的输入，不能从三笔合成数据中恢复。

```python
p=.002
for candidates in [20,500]:print('candidates, threshold, below:',candidates,.05/candidates,p<.05/candidates)
se=net.std(ddof=1)/np.sqrt(3);print('illustrative iid-normal mean interval:',net.mean()+np.array([-1,1])*stats.t.ppf(.975,2)*se)
```
