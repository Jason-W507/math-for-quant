# 样本均值的分布与集中界

重复数据集才能观察抽样分布；一条样本路径不能替代它。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

在每个样本量下独立重复 Bernoulli 和 Cauchy 平均。

```python
repeats=3000;p=.3
for n in [10,100,1000]:
    means=rng.binomial(n,p,size=repeats)/n
    cauchy=rng.standard_cauchy((repeats,n)).mean(axis=1)
    print(n,'Bernoulli quantiles:',np.quantile(means,[.1,.5,.9]),'Cauchy:',np.quantile(cauchy,[.1,.5,.9]))
```

精确二项尾概率与有界性保证采用同一个事件。

```python
n=200;t=.1;k=np.arange(n+1)
tail=stats.binom.pmf(k,n,p)[abs(k/n-p)>=t-1e-12].sum()
print('exact, Chebyshev, Hoeffding:',tail,p*(1-p)/(n*t*t),2*np.exp(-2*n*t*t))
eps=.05;delta=.05;print('sufficient n:',int(np.ceil(np.log(2/delta)/(2*eps**2))))
```
