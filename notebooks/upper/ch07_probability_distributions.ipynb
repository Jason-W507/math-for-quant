# 推前分布与抽样

把均匀随机数切成概率区间，再将随机变量平方。

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
rng = np.random.default_rng(20260907)
np.set_printoptions(precision=5, suppress=True)
```

逆分布抽样先得到离散取值，再按平方后的取值合并概率。

```python
values=np.array([-1.,0.,2.]);prob=np.array([.25,.5,.25]);u=rng.random(20000)
sample=values[np.searchsorted(np.cumsum(prob),u)]
print('mean, second moment, variance:',values@prob,(values**2)@prob,(values**2)@prob-(values@prob)**2)
for value in np.unique(values**2):print('square value:',value,'prob:',prob[values**2==value].sum())
print('sample mean:',sample.mean())
```

连续变换 X² 的两条原像分支一起贡献概率。

```python
x=rng.uniform(-1,1,20000);y=x*x;grid=np.linspace(0,1,100)
plt.plot(grid,np.sqrt(grid),label='analytic CDF');plt.plot(grid,[(y<=t).mean() for t in grid],label='empirical CDF')
plt.xlabel('y');plt.legend();plt.show();print('square median:',np.median(y),'analytic:',.25)
```
